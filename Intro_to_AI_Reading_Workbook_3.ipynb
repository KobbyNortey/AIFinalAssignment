# About the Reading Workbook

The Reading Workbook is designed to motivate and guide students to read extensively on assigned topics. The workbook contains questions for students to answer based on the readings in the assigned text. Students are expected to answer questions in their own words and in their own style of coding. Keep in mind that all reading exercises and some quiz questions will be based off this workbook. Students who thoughtfully answer the questions in this workbook and who do so with understanding should not have a problem answering the related questions that appear in the reading exercises or quizzes.
